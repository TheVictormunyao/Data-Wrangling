```python

#Question4

''' 4: Treatment of Missing Data

The missing values are treated according to the type and meaning of each column. Numerical columns can be treated using statistical measures where appropriate, while categorical, identification and descriptive columns should not be assigned arbitrary numerical values.

| Column           | Appropriate treatment                                                           | Justification                                                                                                                                                                                                                                  |
| ---------------- | ------------------------------------------------------------------------------- | ---------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------- |
|     Date         | No treatment                                                                    | The Date column has no missing values, so no treatment is required.                                                                                                                                                                            |
|     Time         | Leave as missing                                                                | The actual time of a crash cannot reliably be determined from the other data. Filling it with an estimated time could introduce inaccurate information.                                                                                        |
|     Location     | `"Unknown"`                                                                     | Location is categorical information and cannot meaningfully be calculated using a mean or median. `"Unknown"` indicates that the location was not recorded.                                                                                    |
|   Operator       | `"Unknown"`                                                                     | Operator is categorical information. Since the missing operator cannot reliably be inferred, `"Unknown"` preserves the record without inventing information.                                                                                   |
|      Flight      | `"Unknown"`                                                                     | Flight numbers are identifiers rather than numerical measurements. Therefore, mean or median imputation would not be meaningful.                                                                                                               |
|     Route        | `"Unknown"`                                                                     | Route is descriptive information and cannot reliably be calculated from other records. `"Unknown"` indicates that the route was unavailable.                                                                                                   |
|     Type         | Mode or `"Unknown"`                                                             | Type is categorical data. The mode can be used to replace missing categories, although `"Unknown"` avoids assuming that the missing aircraft had the most common type.                                                                         |
|   Registration   | `"Unknown"`                                                                     | Registration is an identifier, so assigning a calculated value would be inappropriate. `"Unknown"` shows that the registration was not available.                                                                                              |
|     cn/In        | `"Unknown"`                                                                     | This is an identification/reference field. A missing identification number cannot meaningfully be calculated from the other records.                                                                                                           |
|     Aboard       | Median                                                                          | Aboard is a numerical count. The median is less affected by unusually large crashes than the mean, making it a suitable estimate for missing values.                                                                                           |
|   Fatalities     | Median                                                                          | Fatalities is a numerical count and contains potentially extreme values. The median provides a more robust estimate than the mean.                                                                                                             |
|     Ground       | 0, if a blank represents no recorded ground fatalities; otherwise leave missing | Ground represents fatalities occurring on the ground. A value of zero is appropriate only if the dataset's missing entries represent no recorded ground fatalities. Otherwise, the missing value should be retained rather than assuming zero. |
|     Summary      | `"Unknown"`                                                                     | Summary is descriptive text, so its missing content cannot reliably be inferred from other records. `"Unknown"` makes the unavailable information explicit.                                                                                    |

**Conclusion:**
No single method is appropriate for every column. The treatment should depend on the data type and meaning of the variable. Numerical counts can use median imputation where an estimate is reasonable, while categorical, identification and descriptive fields should generally use `"Unknown"` or remain missing when their actual values cannot be determined reliably.'''


df.info()
df.isnull().sum()
```

```python
#Question5
fatality_locations = df[['Date', 'Location', 'Aboard','Fatalities']].copy()
fatality_locations.head()
```

```python
#Question6
fatality_locations.loc[fatality_locations['Fatalities'].idxmax(), 'Date']
```

```python
#Question7
(fatality_locations['Fatalities'] == 0).sum()

fatality_locations[fatality_locations['Fatalities'] == 0]
```

```python
fatality_locations[['Aboard', 'Fatalities']]
```

```python
fatality_locations[['Region', 'State/Country']] = (
    fatality_locations['Location'].str.split(',', n=1, expand=True)
)
```

```python
top_100 = fatality_locations.sort_values(
    by ='Fatalities',
    ascending = False
).head(100)

print(top_100)
```

```python
#Question8

fatalities_by_location = (
    fatality_locations
    .groupby('State/Country')['Fatalities']
    .sum()
    .sort_values(ascending=False)
    .head(25) #picks top25
)

fatalities_by_location.plot(
    kind='pie',
    autopct = '%1.1f%%'
)

import matplotlib.pyplot as plt
plt.ylabel('')
plt.title('Distribution of Fatalities by Country/U.S. State')
plt.show()
```
